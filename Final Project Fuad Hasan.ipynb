```python
import pandas as pd
import gradio as gr
from google import genai

client = genai.Client(
    api_key=os.environ["GEMINI_API_KEY"]
)

MODEL_NAME = "gemini-3.5-flash-lite"
```

```python
from google.colab import files

uploaded = files.upload()

filename = list(uploaded.keys())[0]

if filename.endswith(".xlsx"):
    df = pd.read_excel(filename)
elif filename.endswith(".csv"):
    df = pd.read_csv(filename)
else:
    raise ValueError("File harus berupa XLSX atau CSV")

print("File:", filename)
print("Jumlah baris:", len(df))
print("Jumlah kolom:", len(df.columns))

display(df.head())
```

```python
data_text = df.to_csv(index=False)

# Batasi ukuran data untuk prototype
if len(data_text) > 100000:
    data_text = data_text[:100000]

print(data_text[:3000])
```

```python
import time

def ask_ai(question, max_retries=3):

    prompt = f"""
Anda adalah AI Operational Analyst untuk perusahaan pertambangan.

Tugas Anda adalah menganalisis data operasional yang diberikan dan menjawab
pertanyaan user berdasarkan data tersebut.

ATURAN:
1. Gunakan hanya data yang tersedia.
2. Jangan mengarang angka atau fakta.
3. Jika data yang dibutuhkan tidak tersedia, katakan:
   "Data tidak tersedia."
4. Jika perlu melakukan perhitungan, lakukan berdasarkan data.
5. Gunakan Bahasa Indonesia.
6. Jawaban harus ringkas, jelas, dan mudah dipahami oleh management.
7. Jika menemukan masalah, jelaskan berdasarkan data yang tersedia.
8. Jangan membuat asumsi yang tidak didukung data.

DATA OPERASIONAL:
{data_text}

PERTANYAAN USER:
{question}
"""

    for attempt in range(max_retries):

        try:

            response = client.models.generate_content(
                model=MODEL_NAME,
                contents=prompt
            )

            return response.text

        except Exception as e:

            error_message = str(e)

            print(
                f"Percobaan {attempt + 1}/{max_retries} gagal: "
                f"{error_message}"
            )

            # Jika server sedang overloaded
            if "503" in error_message or "UNAVAILABLE" in error_message:

                if attempt < max_retries - 1:
                    wait_time = 5 * (attempt + 1)

                    print(
                        f"Server sedang sibuk. "
                        f"Menunggu {wait_time} detik..."
                    )

                    time.sleep(wait_time)

                else:
                    return (
                        "Gemini sedang mengalami beban tinggi. "
                        "Silakan coba kembali beberapa saat lagi."
                    )

            else:
                return f"Terjadi error: {error_message}"
```

```python
print("Model yang digunakan:", MODEL_NAME)
```

```python
question = """
Berikan analisis singkat terhadap performa operasional
berdasarkan data yang tersedia.
"""

answer = ask_ai(question)

print(answer)
```

```python
def chatbot(message, history):

    try:
        answer = ask_ai(message)
        return answer

    except Exception as e:
        return f"Terjadi error: {str(e)}"
```

```python
demo = gr.ChatInterface(
    fn=chatbot,
    title="⛏️ Mining Operational AI Analyst",
    description=(
        "Upload data operasional melalui notebook, "
        "kemudian tanyakan apa saja terkait data tersebut."
    ),
    textbox=gr.Textbox(
        placeholder="Contoh: Berapa achievement produksi hari ini?",
        container=True
    )
)

demo.launch(share=True)
```
